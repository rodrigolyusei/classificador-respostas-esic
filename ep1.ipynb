# Imports

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score
```

# Dataframe

```python
df = pd.read_excel('./train.xlsx')
```

```python
df.info()
df.describe()
```

# Dummy Majority

```python
vect = CountVectorizer()
X = vect.fit_transform(df["resp_text"].astype(str))
Y = df.clarity
print("X shape: ", X.shape)
print("Y shape: ", Y.shape)
print()

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=123, stratify=Y)
print("x_train x_test shape: ", x_train.shape, x_test.shape)
print("y_train y_test shape: ", y_train.shape, y_test.shape)
print()

clf = DummyClassifier(strategy='most_frequent', random_state = 123, constant = None).fit(x_train, y_train)
print("Model trained")

predicted = clf.predict(x_test)
score = f1_score(y_test,predicted, average='macro')
print("DummyMaj =====> ", score)
```

# Bow RegLog

```python
vect = CountVectorizer()
X = vect.fit_transform(df["resp_text"].astype(str))
Y = df.clarity
print("X shape: ", X.shape)
print("Y shape: ", Y.shape)
print()

x_train, x_test, y_train, y_test = train_test_split(X,Y, test_size=0.2, random_state=123, stratify=Y)
print("x_train x_test shape: ", x_train.shape, x_test.shape)
print("y_train y_test shape: ", y_train.shape, y_test.shape)
print()

clf = LogisticRegression(class_weight='balanced', max_iter=100, random_state=123).fit(x_train, y_train)
print("Model trained")

predicted = clf.predict(x_test)
score = f1_score(y_test,predicted, average='macro')
print("BoW =====> ", score)
```

# BoW TF-IDF RegLog

```python
vect = TfidfVectorizer()
X = vect.fit_transform(df["resp_text"].astype(str))
Y = df.clarity
print("X shape: ", X.shape)
print("Y shape: ", Y.shape)
print()

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=123, stratify=Y)
print("x_train x_test shape: ", x_train.shape, x_test.shape)
print("y_train y_test shape: ", y_train.shape, y_test.shape)
print()

clf = LogisticRegression(class_weight='balanced', max_iter=100, random_state=123).fit(x_train, y_train)
print("Model trained")

predicted = clf.predict(x_test)
score = f1_score(y_test,predicted, average='macro')
print("BoW TF-IDF =====> ", score)
```
