# Imports

```python
import warnings
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.feature_selection import VarianceThreshold, SelectKBest
from sklearn.linear_model import LogisticRegression
from sklearn.exceptions import ConvergenceWarning
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, classification_report, confusion_matrix
```

```python
# "default" to show warnings, "ignore" to hide warnings
warnings.filterwarnings("ignore", category=ConvergenceWarning)
```

# Dataframe

```python
df = pd.read_excel('./train.xlsx')
df.describe()
```

# Dummy Majority

```python
X = df["resp_text"].astype(str)
Y = df["clarity"]
print("X Y: ", X.shape, Y.shape)
print()

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=123, stratify=Y)
print("x_train x_test: ", x_train.shape, x_test.shape)
print("y_train y_test: ", y_train.shape, y_test.shape)
print()

vect = CountVectorizer()
x_train_vec = vect.fit_transform(x_train)
x_test_vec = vect.transform(x_test)
print("x_train_vec x_test_vec: ", x_train_vec.shape, x_test_vec.shape)
print()

clf = DummyClassifier(strategy='most_frequent', random_state = 123, constant = None).fit(x_train_vec, y_train)
print("Model trained")

predicted = clf.predict(x_test_vec)
score = f1_score(y_test, predicted, average='macro')
print("DummyMaj =====> ", score)
print("DummyMaj report:")
print(classification_report(y_test, predicted, zero_division=0))
print("DummyMaj confusion matrix:")
print(confusion_matrix(y_test, predicted, labels=["c1", "c234", "c5"]))
```

# Bow RegLog

```python
X = df["resp_text"].astype(str)
Y = df["clarity"]
print("X Y: ", X.shape, Y.shape)
print()

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=123, stratify=Y)
print("x_train x_test: ", x_train.shape, x_test.shape)
print("y_train y_test: ", y_train.shape, y_test.shape)
print()

vect = CountVectorizer()
x_train_vec = vect.fit_transform(x_train)
x_test_vec = vect.transform(x_test)
print("x_train_vec x_test_vec: ", x_train_vec.shape, x_test_vec.shape)
print()

clf = LogisticRegression(class_weight='balanced', max_iter=100, random_state=123).fit(x_train_vec, y_train)
print("Model trained")

predicted = clf.predict(x_test_vec)
score = f1_score(y_test, predicted, average='macro')
print("BoW =====> ", score)
print("BoW report:")
print(classification_report(y_test, predicted))
print("BoW confusion matrix:")
print(confusion_matrix(y_test, predicted, labels=["c1", "c234", "c5"]))
```

# BoW TF-IDF kbest fixed RegLog

```python
X = df["resp_text"].astype(str)
Y = df["clarity"]
print("X Y: ", X.shape, Y.shape)
print()

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=123, stratify=Y)
print("x_train x_test: ", x_train.shape, x_test.shape)
print("y_train y_test: ", y_train.shape, y_test.shape)
print()

vect = TfidfVectorizer()
x_train_vec = vect.fit_transform(x_train)
x_test_vec = vect.transform(x_test)
print("x_train_vec x_test_vec: ", x_train_vec.shape, x_test_vec.shape)

# Remove constant features using training data only
variance_filter = VarianceThreshold(threshold=0.0)
x_train_filt = variance_filter.fit_transform(x_train_vec)
x_test_filt = variance_filter.transform(x_test_vec)
print("x_train_filt x_test_filt filtered: ", x_train_filt.shape, x_test_filt.shape)

kvalue = min(5000, x_train_filt.shape[1])
best_sel = SelectKBest(k=kvalue)
best_fit = best_sel.fit(x_train_filt, y_train)
x_train_best = best_fit.transform(x_train_filt)
x_test_best = best_fit.transform(x_test_filt)
print("x_train_best x_test_best:", x_train_best.shape, x_test_best.shape)
print()

clf = LogisticRegression(class_weight="balanced", max_iter=100, random_state=123).fit(x_train_best, y_train)
print("Model trained")

predicted = clf.predict(x_test_best)
score = f1_score(y_test, predicted, average="macro")
print("kvalue: ", kvalue)
print("BoW TF-IDF kbest =====> ", score)
print("BoW TF-IDF kbest report:")
print(classification_report(y_test, predicted))
print("BoW TF-IDF kbest confusion matrix:")
print(confusion_matrix(y_test, predicted, labels=["c1", "c234", "c5"]))
```

# BoW TF-IDF kbest searched RegLog

```python
X = df["resp_text"].astype(str)
Y = df["clarity"]
print("X Y: ", X.shape, Y.shape)
print()

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=123, stratify=Y)
print("x_train x_test: ", x_train.shape, x_test.shape)
print("y_train y_test: ", y_train.shape, y_test.shape)
print()

vect = TfidfVectorizer()
x_train_vec = vect.fit_transform(x_train)
x_test_vec = vect.transform(x_test)
print("x_train_vec x_test_vec: ", x_train_vec.shape, x_test_vec.shape)

# Remove constant features using training data only
variance_filter = VarianceThreshold(threshold=0.0)
x_train_filt = variance_filter.fit_transform(x_train_vec)
x_test_filt = variance_filter.transform(x_test_vec)
print("x_train_filt x_test_filt filtered: ", x_train_filt.shape, x_test_filt.shape)
print()

bestScore = 0.0
bestK = 0
kvalues = list(range(x_train_filt.shape[1], 10000, -10000))
for kvalue in kvalues:
    best_sel = SelectKBest(k=kvalue)
    best_fit = best_sel.fit(x_train_filt, y_train)
    x_train_best = best_fit.transform(x_train_filt)
    x_test_best = best_fit.transform(x_test_filt)

    clf = LogisticRegression(class_weight="balanced", max_iter=100, random_state=123).fit(x_train_best, y_train)
    print("Model trained")

    predicted = clf.predict(x_test_best)
    score = f1_score(y_test, predicted, average="macro")
    print("kvalue: ", kvalue)
    print("BoW TF-IDF kbest =====> ", score)
    print()
    
    if score > bestScore:
        bestScore = score
        bestK = kvalue

print("bestScore = ", bestScore)
print("bestK = ", bestK)
```

# NGram TF-IDF RegLog

```python
X = df["resp_text"].astype(str)
Y = df["clarity"]
print("X Y: ", X.shape, Y.shape)
print()

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=123, stratify=Y)
print("x_train x_test: ", x_train.shape, x_test.shape)
print("y_train y_test: ", y_train.shape, y_test.shape)
print()

vect = TfidfVectorizer(ngram_range=(3, 6), analyzer='char', min_df=3, sublinear_tf=True)
x_train_vec = vect.fit_transform(x_train)
x_test_vec = vect.transform(x_test)
print("x_train_vec x_test_vec: ", x_train_vec.shape, x_test_vec.shape)
print()

clf = LogisticRegression(class_weight='balanced', max_iter=1000, solver='saga', random_state=123).fit(x_train_vec, y_train)
print("Model trained")

predicted = clf.predict(x_test_vec)
score = f1_score(y_test, predicted, average='macro')
print("NGram TF-IDF =====> ", score)
print("NGram TF-IDF report:")
print(classification_report(y_test, predicted))
print("NGram TF-IDF confusion matrix:")
print(confusion_matrix(y_test, predicted, labels=["c1", "c234", "c5"]))
```
